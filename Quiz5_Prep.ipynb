```python
# Create a new column of the DataFrame called "tokens" where each string in the reviews column is converted to a list of individual lower-case words.

df["tokens"] = df["review"].str.findall(r"[A-Za-z]+").apply(lambda x: [w.lower() for w in x])
```

```python
# Unravel the tokens column into a single Pandas Series containing all the words appearing in the all the reviews. 
# Create a new Pandas series with the probability distribution of the words appearing the in the DataSet. 
# Then create a bar plot of the probabilities for a subset of the most common words.

words = df["tokens"].explode()
probs = words.value_counts(normalize=True)
probs.head(20).plot(kind="bar")

```
