```python
# 1
# You will be given a code cell that calculates the coefficients of a quadratic model (y = a * x^2 + b * x + c) between two variables. All code in this cell will be correct.
# Then in the next code cell, a figure is generated with a scatter plot of the individual data and a line plot of the model prediction. 
# However, the plot is obviously wrong. Make necessary modifications to generate a correct plot.

import numpy as np
import matplotlib.pyplot as plt

# Example data
x = np.linspace(-10, 10, 50)
y = 2 * x**2 + 3 * x + 5 + np.random.randn(50) * 10

# Fit quadratic model
coeffs = np.polyfit(x, y, 2)
a, b, c = coeffs
y_pred = a * x**2 + b * x + c

# FIX: Sort x before plotting the line for correct shape
order = np.argsort(x)
x_sorted = x[order]
y_pred_sorted = y_pred[order]

# Plot
plt.scatter(x, y, color='blue', label='Data')
plt.plot(x_sorted, y_pred_sorted, color='red', label='Model Prediction')
plt.title('Quadratic Fit: y = a*x^2 + b*x + c')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

```python
# 3
# Read in a .csv file to create a DataFrame which contains conversion rates for Euros to U.S. dollars and British Pounds to U.S. dollars over time. 
# Compute non-dimensional percentage growth values for each and plot together on a single plot. 
# (You can use matplotlib.pyplot or the .plot() method connected to a DataFrame).


import pandas as pd

# Read CSV file
df_conv = pd.read_csv('currency_rates.csv')  # Columns: ['Date', 'Euro_USD', 'GBP_USD']
df_conv['Date'] = pd.to_datetime(df_conv['Date'])

# Compute non-dimensional percentage growth
df_conv['Euro_growth'] = df_conv['Euro_USD'] / df_conv['Euro_USD'].iloc[0] * 100
df_conv['GBP_growth'] = df_conv['GBP_USD'] / df_conv['GBP_USD'].iloc[0] * 100

# Plot both on same figure
df_conv.plot(x='Date', y=['Euro_growth', 'GBP_growth'], title='Currency Growth Comparison')
```

```python
# 4
# Continuing with the DataFrame of conversion rates, create a new column of Boolean type, which checks whether the 
# conversion rate for one of the currencies was greater than a given value.

# Example threshold
threshold = 1.20

# Create Boolean column
df_conv['High_Euro'] = df_conv['Euro_USD'] > threshold

print(df_conv.head())
```

```python
# 5
# Continuing Problem 4, make a line plot of the conversion rate for this currency and highlight regions where the conversion 
# rate is larger than a given value, using the function "plt.fill_between"

import matplotlib.pyplot as plt

plt.plot(df_conv['Date'], df_conv['Euro_USD'], label='Euro/USD', color='blue')
plt.fill_between(df_conv['Date'], df_conv['Euro_USD'],
                 where=df_conv['Euro_USD'] > threshold,
                 color='red', alpha=0.3, label='Above Threshold')
plt.title('Euro/USD Conversion Rate Highlighted')
plt.xlabel('Date')
plt.ylabel('Rate')
plt.legend()
plt.show()
```

```python
# 6
# A code cell creates a DataFrame with the following columns. All columns are numerical:
# time: A column containing evenly spaced values between 0 and 2pi. Denote them by t0, t1, ..., tN 
# sin: A column containing the sine function evaluated at those points. I.e. sin(t0), sin(t1), ..., sin(tN)
# cos: A column containing the cosine function evaluated at those points. I.e.cos(t0), cos(t1), ..., cos(tN)

# Using the ".diff()" method, create a new column called "d_sin_dt" with the following formula:    (sin(tk+1) - sin(tk) / tk+1 - tk)
# From calculus, this should be close to cos(tk), so plot (time vs. cos) and (time vs. d_sin_dt) together on a single figure to it.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Create time data
t = np.linspace(0, 2 * np.pi, 100)
df = pd.DataFrame({
    'time': t,
    'sin': np.sin(t),
    'cos': np.cos(t)
})

# Compute numerical derivative of sin using diff
df['d_sin_dt'] = df['sin'].diff() / df['time'].diff()

# Plot
plt.plot(df['time'], df['cos'], label='cos(t)')
plt.plot(df['time'], df['d_sin_dt'], label='d_sin_dt', linestyle='--')
plt.title('Comparing cos(t) and d_sin_dt')
plt.xlabel('time')
plt.ylabel('value')
plt.legend()
plt.show()


```
